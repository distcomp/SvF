```python
import numpy as np
import sympy as sp
```

```python
F00, F01, F10, F11, la, mu = sp.symbols("F00, F01, F10, F11, la, mu")
Fxy = (1-mu)*((1-la)*F00 + la*F10) + mu*((1-la)*F01 + la*F11)
Fyx = (1-la)*((1-mu)*F00 + mu*F01) + la*((1-mu)*F10 + mu*F11)
(Fxy - Fyx).simplify()
```

```python
F000, F010, F100, F110, F001, F011, F101, F111, w1, w2, w3 = sp.symbols("F000, F010, F100, F110, F001, F011, F101, F111, w1, w2, w3")
Fxy0 = (1-w2)*((1-w1)*F000 + w1*F100) + w2*((1-w1)*F010 + w1*F110)
Fxy1 = (1-w2)*((1-w1)*F001 + w1*F101) + w2*((1-w1)*F011 + w1*F111)
Fxyz = (1-w3)*Fxy0 + w3*Fxy1
print(Fxy0)
print(Fxy1)
print(Fxyz.simplify())
F0yz = (1-w3)*((1-w2)*F000 + w2*F010) + w3*((1-w2)*F001 + w2*F011)
F1yz = (1-w3)*((1-w2)*F100 + w2*F110) + w3*((1-w2)*F101 + w2*F111)
Fyzx = (1-w1)*F0yz + w1*F1yz
print((Fxyz - Fyzx).simplify())
```

```python
a0, a1, b0, b1, c0, c1 = sp.symbols("a0, a1, b0, b1, c0, c1")
Fxy0n = b0*(a0*F000 + a1*F100) + b1*(a0*F010 + a1*F110)
Fxy1n = b0*(a0*F001 + a1*F101) + b1*(a0*F011 + a1*F111)
Fxyzn = c0*Fxy0n + c1*Fxy1n
print(Fxyzn.simplify())
print(Fxyzn.simplify().expand())
```
